```python
#I. define the dataset codes to download
from apiEurostat.transform_reponse import create_lookup_tables


dataset_codes = [
    "env_wasmun",
    "env_wasobl",
]

for datasetCode in dataset_codes:
    print(f"Downloading data for dataset code: {datasetCode}")
    #1.download the data for the defined dataset codes
    data = download_eurostat(datasetCode)
    #2. create a pandas dataframe from the downloaded data
    df = tr.response_to_dataframe(data)
    #3. create look up tables
    lookups = create_lookup_tables(data)
    #4. get source code and source label  
    #source_code = tr.get_source_code(data)
    source_label = tr.get_source_label(data)
    #5. create directory to save the data and lookups
    output_dir = tr.create_output_directory(datasetCode)

    #save the dataframe, lookups, and source label to parquet files in the output directory
    outputs = {
        "df": df,
        "lookups": lookups,
        "source_label": source_label
    }

    for name, data in outputs.items():
        tr.save_parquet(data, output_dir, name)



```
