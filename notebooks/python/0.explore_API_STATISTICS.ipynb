```python
#See guidelines:
# https://ec.europa.eu/eurostat/web/user-guides/data-browser/api-data-access/api-detailed-guidelines/api-statistics
#Filters are optional and can be any dimension present in the data. Filters start with a question mark ("?") and are separated by an ampersand (&) if there are several filters.
#For time: both "time and"time_period" are accepted. also "utnilTilePeriod", "sinceTimePeriod", and "lastTimePeriod".
#geoLevel: allows retreiving a lisg of GEO codes according to ther level in the NUTS classification: EU15, EU25, EU27, ANY COUNTRY CODE.

import requests
import pandas as pd
from pyjstat import pyjstat


```

```python
#API configuration
host_Url =  "https://ec.europa.eu/eurostat/api/dissemination/"
service = "statistics"
version = "1.0"
response_type = "data"
#dynamic part    
datasetCode = "ENV_WASMUN"
#Query  parameters
params = {
   
    "format": "JSON",
    "lang": "EN",
    "sinceTimePeriod": "2010",
    'untilTimePeriod': '2026',
    'geoLevel': 'country'

}

url = f"{host_Url}{service}/{version}/{response_type}/{datasetCode}"

```

```python
#Execute request
response = requests.get(url,params=params)
#explore response
print(response.url)
response.status_code
```

```python


#explore response content
data = response.json()
for key in data.keys():
    print(f"{key}: {data[key]}")


#id: dimensions/variables in the data
#size: number of categories within each dimension
#dimension: the value of the categories within each diemension (is the dictionary)
#value: the actual value of the indicator


#data['dimension']['geo']['category']['index'] #the index of the categories within the dimension (MAKE A FUNCTION TO GET THE CATEGORIES AND THEIR INDEXES)
```
