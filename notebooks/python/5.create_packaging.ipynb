```python

import pandas as pd
from pathlib import Path
import json


ROOT = Path.cwd().parents[1]     
DATA_SOURCE = 'env_waspacr'
DATA_RAW_DIR = ROOT / "data/raw" / DATA_SOURCE     # notebooks/python -> project root
EXPORT_DIR = ROOT / "data/interim" 
EXPORT_DIR.mkdir(parents=True, exist_ok=True)
EXPORT_FILE = EXPORT_DIR / f"{DATA_SOURCE}.parquet"
df = pd.read_parquet(DATA_RAW_DIR / "df.parquet")
codebook = json.load(open(DATA_RAW_DIR / "lookup.json"))

df.head()


    
```

```python
#check the value geo is DE and wst_oper is GEN
#use the codebook to create a new column called waste_name that maos the waste column with the corresponding waste name
df['waste_name'] = df['waste'].map(codebook['waste'])
#df[(df['geo'] == 'DE') & (df['unit'] == 'RT_TGT2025') & (df['time'] == "2023")]
#select only the relevant columns for further analysis in 2023 and unit RT_TGT2025
df_final = df[(df['time'] == "2023") & (df['unit'] == 'RT_TGT2025')][['geo', 'time', 'waste_name', 'value']]

#count missing values of value by country
df_final.groupby('geo')['value'].apply(lambda x: x.isna().sum())
```
